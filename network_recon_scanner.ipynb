```python
"""
Module: network_recon_scanner.py
Author: Portfolio Project - IIT Kanpur B.Cyber Admissions
Description: A lightweight network reconnaissance utility utilizing TCP socket connections
             to audit open services on a targeted host machine.
"""

import socket
import sys
from datetime import datetime

def banner():
    print("=" * 60)
    print("          ADVANCED TCP PACKET AUDIT & RECONNAISSANCE TOOL          ")
    print("=" * 60)

def run_scanner():
    banner()

    # User Input with automated protocol stripping for stability
    raw_target = input("[*] Enter Target Domain or IP (e.g., scanme.nmap.org): ").strip()

    # Sanitizing input strings (stripping web structures if pasted)
    target_host = raw_target.replace("https://", "").replace("http://", "").split("/")[0]

    if not target_host:
        print("[-] Critical Error: Empty target field.")
        return

    # DNS Resolution
    print(f"\n[*] Resolving hostname target: {target_host}...")
    try:
        target_ip = socket.gethostbyname(target_host)
    except socket.gaierror:
        print(f"[-] Resolution Failure: Domain '{target_host}' could not be resolved.")
        sys.exit()

    # Scanner Metrics Banner
    print("-" * 60)
    print(f"Target IPv4 Address : {target_ip}")
    print(f"Scan Initialization : {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    print("-" * 60)

    # Core Common Ports List (SSH, SMTP, HTTP, HTTPS, RDP)
    ports_to_scan = [22, 25, 80, 443, 3389]
    open_ports = []

    # Network Scanning Core Loop
    for port in ports_to_scan:
        # Initializing AF_INET (IPv4) and SOCK_STREAM (TCP) socket
        s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

        # 2.0-second network latency safety boundary
        s.settimeout(2.0)

        # connect_ex returns an error code instead of throwing an unhandled crash
        result = s.connect_ex((target_ip, port))

        if result == 0:
            print(f"[+] Port {port:<5} | STATE: OPEN     | Handshake Successful")
            open_ports.append(port)
        else:
            print(f"[-] Port {port:<5} | STATE: CLOSED   | Handshake Refused")

        s.close()

    # Final Execution Metrics Summary
    print("-" * 60)
    print(f"[+] Scan Complete. Identified {len(open_ports)} open entry points.")
    print("=" * 60)

if __name__ == "__main__":
    try:
        run_scanner()
    except KeyboardInterrupt:
        print("\n[-] Operation forcefully terminated by user interface interrupt (Ctrl+C).")
        sys.exit()
```
