```python
"""
Module: network_crypto_engine.py
Author: Portfolio Project - IIT Kanpur B.Cyber Admissions
Description: A functional symmetric cryptography processing utility demonstrating bitwise
             data obfuscation and cryptographic confidentiality algorithms.
"""

import sys

def cryptographic_transform(input_string, key_character):
    """
    Applies a mathematical bitwise XOR transformation across a string block.
    In symmetric cryptography, passing the ciphertext back through the exact same
    key reveals the original plaintext message.
    """
    transformed_output = ""

    # Process the message string character-by-character
    for char in input_string:
        # ord(char) extracts the numerical ASCII value of the letter
        # ord(key_character) extracts the numerical ASCII value of the secret key
        # The '^' operator performs a bitwise XOR gate calculation
        bitwise_result = ord(char) ^ ord(key_character)

        # chr() turns the transformed numerical value back into a character string
        transformed_output += chr(bitwise_result)

    return transformed_output

def run_crypto_suite():
    print("=" * 60)
    print("        SYMMETRIC PAYLOAD CRYPTOGRAPHY & DATA PROTECTION        ")
    print("=" * 60)

    # Secure User Input Configuration
    secret_message = input("[*] Enter sensitive plaintext payload data: ").strip()
    secret_key = input("[*] Define a single-character secret key (e.g., 'X'): ").strip()

    # Input validation gate to enforce cryptographic key standards
    if len(secret_key) != 1:
        print("[-] Algorithmic Constraint Error: Secret key must be exactly ONE character.")
        sys.exit()

    print("\n" + "-" * 60)
    print("INITIALIZING ENCRYPTION CYCLE...")
    print("-" * 60)

    # 1. Execute Encryption Algorithm
    ciphertext = cryptographic_transform(secret_message, secret_key)
    print(f"[+] Plaintext Input : {secret_message}")
    print(f"[+] Ciphertext Output: {ciphertext.encode('utf-8', errors='replace')} (Byte Obfuscated)")

    print("\n" + "-" * 60)
    print("INITIALIZING DECRYPTION CYCLE (Simulating Recipient)...")
    print("-" * 60)

    # 2. Execute Decryption Algorithm (Symmetric reversal)
    decrypted_plaintext = cryptographic_transform(ciphertext, secret_key)
    print(f"[+] Ciphertext Input : {ciphertext}")
    print(f"[+] Decrypted Output : {decrypted_plaintext}")

    # 3. Verification Integrity Gate
    print("-" * 60)
    if secret_message == decrypted_plaintext:
        print("[✅ INTEGRITY MATCH] Decrypted payload exactly mirrors plaintext data layer.")
    else:
        print("[❌ FAILURE] Cryptographic drift detected. Packet corrupted.")
    print("=" * 60)

if __name__ == "__main__":
    run_crypto_suite()
```
