```python
"""
Module: log_intrusion_detector.py
Author: Portfolio Project - IIT Kanpur B.Cyber Admissions
Description: A security metrics script that reads server access logs, counts authentication
             failures, and flags anomalous Brute-Force activity metrics.
"""

def analyze_logs(log_file_path, failure_threshold=3):
    print("=" * 60)
    print("          SYSTEM AUTOMATION LOG ANALYSIS & INTRUSION DETECTION          ")
    print("=" * 60)
    print(f"[*] Parsing log target: {log_file_path}")
    print(f"[*] Trigger Flag Threshold: > {failure_threshold} failed authentication attempts\n")

    failed_attempts = {}

    try:
        with open(log_file_path, "r") as file:
            for line in file:
                # We look for lines containing authentication anomalies
                if "STATUS: FAILED" in line:
                    # Cleanly isolate the IP address from the log string text pattern
                    parts = line.split(" - ")
                    ip_part = parts[1] # "IP: 192.168.1.50"
                    ip_address = ip_part.replace("IP: ", "").strip()

                    # Track failure occurrences dynamically inside a dictionary map
                    failed_attempts[ip_address] = failed_attempts.get(ip_address, 0) + 1

    except FileNotFoundError:
        print(f"[-] Evaluation Error: Target log file system pattern '{log_file_path}' not found.")
        return

    # Evaluate the tracking dictionary map for anomalies exceeding threshold boundaries
    incident_triggered = False
    print("-" * 60)
    print(f"{'SUSPECT IP ADDRESS':<25} | {'FAILED ATTEMPTS':<15} | {'SECURITY STATUS'}")
    print("-" * 60)

    for ip, count in failed_attempts.items():
        if count >= failure_threshold:
            status = "⚠️ ALERT: BRUTE-FORCE DETECTED"
            incident_triggered = True
        else:
            status = "✅ PASS: Normal Volatility"

        print(f"{ip:<25} | {count:<15} | {status}")

    print("-" * 60)
    if not incident_triggered:
        print("[+] Threat Matrix clean. No anomalous credential filling detected.")
    else:
        print("[!] Risk Assessment complete. Immediate incident triage response recommended.")
    print("=" * 60)

if __name__ == "__main__":
    analyze_logs("server_access.log", failure_threshold=3)
```
