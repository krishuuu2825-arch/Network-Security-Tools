```python
import socket
import sys
from datetime import datetime

# Define a safe public target provided by Nmap for legal scanner testing
target_host = "scanme.nmap.org"

try:
    target_ip = socket.gethostbyname(target_host)
except socket.gaierror:
    print("\n[-] Hostname could not be resolved.")
    sys.exit()

print("-" * 50)
print(f"Scanning Target: {target_ip} ({target_host})")
print(f"Time Started: {str(datetime.now())}")
print("-" * 50)

# Fixed: Explicitly defining standard ports to scan (80 = HTTP, 443 = HTTPS, 22 = SSH)
ports_to_scan = [22, 80, 443]

for port in ports_to_scan:
    s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    s.settimeout(2.0) # Slightly higher timeout for cloud network latency
    result = s.connect_ex((target_ip, port))

    if result == 0:
        print(f"[+] Port {port}: OPEN")
    else:
        print(f"[-] Port {port}: Closed/Filtered")
    s.close()
```
